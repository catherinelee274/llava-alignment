Let 
N be the number of parameters, 
P the precision. 


The following components will have to fit together in memory:

Model to train: N×P

Reference model: the reference model is the same as the model to train, so it also requires N×P
Gradients: we train the whole model, and each parameter requires a gradient, so it requires N×P
Optimizer states: we use AdamW, which requires two states per parameter, so it requires 2 x N x P
